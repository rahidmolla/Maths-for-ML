```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

np.random.seed(42)

data = np.random.normal(
    loc=50,
    scale=10,
    size=10000
)

print(data[:10])
```

```python
print("Mean:", np.mean(data))
print("Median:", np.median(data))
print("Standard deviation:", np.std(data))
```

```python
plt.figure(figsize=(10, 5))

sns.histplot(
    data,
    bins=50,
    kde=True
)

plt.title("Normal Distribution")
plt.xlabel("Value")
plt.ylabel("Frequency")

plt.show()
```

```python
import numpy as np

def normal_pdf_numpy(x, mean, std):
    first_part = 1 / (std * np.sqrt(2 * np.pi))

    exponent = -0.5 * ((x - mean) / std) ** 2

    return first_part * np.exp(exponent)

  x = np.linspace(30, 110, 500)

pdf_values = normal_pdf_numpy(
    x,
    mean=70,
    std=10
)
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(10, 5))

plt.plot(x, pdf_values)

plt.title("Normal Distribution PDF")
plt.xlabel("x")
plt.ylabel("Probability Density")

plt.grid()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
import math

def normal_cdf(x, mean, std):
    z = (x - mean) / (std * math.sqrt(2))

    return 0.5 * (1 + math.erf(z))

x = np.linspace(30, 110, 500)

cdf_values = [
    normal_cdf(value, mean=70, std=10)
    for value in x
]

plt.figure(figsize=(10, 5))

plt.plot(x, cdf_values)

plt.title("Normal Distribution CDF")
plt.xlabel("Value")
plt.ylabel("Cumulative Probability")

plt.grid()
plt.show()
```

```python
import numpy as np
from scipy import stats

data = np.random.normal(
    loc=70,
    scale=10,
    size=100000
)

print("Mean:", np.mean(data))
print("Median:", np.median(data))
print("Mode:", stats.mode(data, keepdims=False).mode)
```
