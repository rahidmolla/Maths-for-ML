```python
import numpy as np

def five_number_summary(data):
    data = np.asarray(data)

    if data.size == 0:
        raise ValueError("Dataset cannot be empty")

    q1, median, q3 = np.percentile(
        data, [25, 50, 75]
    )

    return {
        "Minimum": np.min(data),
        "Q1": q1,
        "Median": median,
        "Q3": q3,
        "Maximum": np.max(data)
    }

data = [12, 15, 18, 20, 22, 25, 28, 30, 35]

print(five_number_summary(data))
```

```python
from scipy.stats import zscore

data = [10, 20, 30, 40, 50, 60, 70, 1000]

z_scores = zscore(data)

print(z_scores)
outliers = np.array(data)[np.abs(z_scores) > 3]

print(outliers)
```

```python
import pandas as pd

rolls = [1, 2, 3, 4, 2, 6, 5, 3, 2, 1,
         4, 6, 2, 5, 3, 4, 1, 2, 6, 3]

pmf = pd.Series(rolls).value_counts().sort_index()

pmf = pmf / len(rolls)

print(pmf)
```

```python
cdf = np.cumsum(pmf)

print(cdf)
```

```python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(-1, 3, 400)

y = [cdf(value) for value in x]

plt.plot(x, y)
plt.xlabel("x")
plt.ylabel("F(x)")
plt.title("CDF of Uniform Distribution")
plt.grid(True)
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm

data = np.array([10, 12, 13, 15, 20])

mu, sigma = norm.fit(data)

x = np.linspace(5, 25, 300)
y = norm.pdf(x, loc=mu, scale=sigma)

plt.hist(data, bins=5, density=True, alpha=0.5)
plt.plot(x, y)

plt.xlabel("Value")
plt.ylabel("Density")
plt.title("Data and Fitted Normal Distribution")
plt.show()
```

```python
from scipy.stats import gaussian_kde

data = np.array([1, 2, 3])

kde = gaussian_kde(data)

x_values = np.linspace(0, 4, 100)
density_values = kde(x_values)

plt.plot(x_values, density_values)
plt.scatter(data, [0] * len(data))

plt.xlabel("x")
plt.ylabel("Density")
plt.title("SciPy Gaussian KDE")
plt.show()
```
