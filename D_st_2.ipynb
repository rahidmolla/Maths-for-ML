```python
import numpy as nimport pandas as pd

df = pd.DataFrame({
    'Student': ['A', 'B', 'C', 'D', 'E'],
    'Marks': [40, 60, 70, 80, 100]
})
df['Percentile_Rank'] = df['Marks'].rank(
    pct=True,
    method='average'
) * 100

print(df)
```

```python
import numpy as np

def five_number_summary(data):
    data = np.asarray(data)

    if data.size == 0:
        raise ValueError("Dataset cannot be empty")

    q1, median, q3 = np.percentile(
        data, [25, 50, 75]
    )

    return {
        "Minimum": np.min(data),
        "Q1": q1,
        "Median": median,
        "Q3": q3,
        "Maximum": np.max(data)
    }

data = [12, 15, 18, 20, 22, 25, 28, 30, 35]

print(five_number_summary(data))
```

```python
from scipy.stats import zscore

data = [10, 20, 30, 40, 50, 60, 70, 1000]

z_scores = zscore(data)

print(z_scores)
outliers = np.array(data)[np.abs(z_scores) > 3]

print(outliers)
```
